```python
import sys
sys.path.append('/home/GUO_Zimeng/coding/data_cleaning_aug_rebuild/module_based_dialogue_synthesis/tools/deduplicate_assistant.py')

from deduplicate_assistant import process
import json
from pathlib import Path
import copy

```

```python
INPUT_FILE = "/home/GUO_Zimeng/coding/data_cleaning_aug_rebuild/module_based_dialogue_synthesis/tool_analysis/data/data-backbone-due-2w-variants-260723.json"
OUTPUT_FILE = "/home/GUO_Zimeng/coding/data_cleaning_aug_rebuild/module_based_dialogue_synthesis/tool_analysis/result/data-backbone-due-2w-variants-260723_0.6v1.json"

# Cell 3: 读取数据
with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    raw_data = json.load(f)

print(f"✅ 原始对话数: {len(raw_data)}")

data_copy = copy.deepcopy(raw_data)
processed, stats = process(data_copy, threshold=0.85, ignore_numbers=True, return_stats=True)

print(f"总对话数: {stats['total_dialogues']}")
print(f"存在重复的对话数: {stats['duplicate_dialogues']}")
print(f"实际修改了 messages 的对话数: {stats['modified_dialogues']}")
print(f"删除的轮对数 (user+assistant): {stats['removed_pairs']}")

print(f"✅ 处理后对话数: {len(processed)}")

# Cell 5: 保存结果
output_path = Path(OUTPUT_FILE)
output_path.parent.mkdir(parents=True, exist_ok=True)
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(processed, f, ensure_ascii=False, indent=2)

print(f"✅ 结果已保存至: {output_path}")
```
